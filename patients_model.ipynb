```python
import random
import pandas as pd
from datetime import datetime, timedelta
from google.colab import files
```

```python
patient_data = []
visit_dt = datetime.now()
today = datetime.now()
```

```python
# สุ่มทุกอย่าง ชื่อ วันเกิด เบอร์ อาการ ยา โรคประจำตัว
for i in range(1, 301):
    # รหัส
    patient_id = f"P{i:03d}"
    # ชื่อ
    fullname = f"{random.choice(first_names)} {random.choice(last_names)}"
    # วันเกิด
    dob_dt = today - timedelta(days=random.randint(5 * 365, 90 * 365))
    dob = dob_dt.strftime("%Y-%m-%d")
    # อายุ
    age = today.year - dob_dt.year - ((today.month, today.day) < (dob_dt.month, dob_dt.day))
    # เบอร์โทร
    phone_number = f"08{random.randint(10000000, 99999999)}"
    # แพ้ยา
    allergies = random.choice(allergies_list)
    # โรคประจำตัว
    diseases = random.choice(diseases_list)
    # สี
    urgency_level = random.choice(list(symptoms_pool.keys()))
    # อาการ
    symptom_detail = random.choice(symptoms_pool[urgency_level])

    patient_data.append({
        "รหัสผู้ป่วย": patient_id,
        "ชื่อ - นามสกุล": fullname,
        "วันเกิด": dob,
        "อายุ": age,
        "เบอร์โทรศัพท์": phone_number,
        "ประวัติแพ้ยา": allergies,
        "โรคประจำตัว": diseases,
        "ระดับอาการ": urgency_level,
        "อาการ": symptom_detail,
})
```

```python
df_patients = pd.DataFrame(patient_data)
df_patients.to_csv("patients_300.csv", index=False, encoding="utf-8-sig")

print("สร้างข้อมูลผู้ป่วยครบ 300 คนเรียบร้อยแล้ว!")
files.download("patients_300.csv")
```

```python
medical_records = []

# สุ่ม
for i in range(1, 1001):
    # สุ่มคนไข้ + หมอ
    patient = random.choice(patients_list)
    doctor = random.choice(doctors_data)

    # ดึงอาการของผู้ป่วยคนนั้นมาเช็ก
    symptom = patient.get("อาการ", patient.get("อาการที่มา", "ไอ มีน้ำมูก"))

    # แมปปิ้งผลวินิจฉัย/ยา/สถานะ/สีคัดกรอง จากอาการจริง
    rules_dict = medical_recordsl_rules
    if symptom in rules_dict:
        rule = rules_dict[symptom]
        meds = random.choice(rule["meds_list"])
        status = random.choice(rule["status_list"])
    else:
        meds = "Paracetamol 500mg"
        status = "หายขาด"

    # The treatment date should be based on 'today's' date, not a random day in the past.
    treatment_date = today
    random_hours = random.randint(8, 16)
    random_minutes = random.randint(0, 59)
    random_seconds = random.randint(0, 59)
    treatment_time = treatment_date.replace(
        hour=random_hours, minute=random_minutes, second=random_seconds
    ).strftime("%Y-%m-%d %H:%M:%S")

    # 4. บันทึกข้อมูลพร้อมคอลัมน์ "เวลารักษา"
    medical_records.append({
        "รหัสรักษา": f"TR{i:03d}",
        "เวลารักษา": treatment_time,
        "รหัสผู้ป่วย": patient.get("รหัสผู้ป่วย") or patient.get("P_id"),
        "รหัสหมอ": doctor["D_id"],
        "ชื่อหมอ": doctor["doctor_name"],
        "อาการที่มา": symptom,
        "ยาที่จะสั่ง": meds,
        "สถานะรักษา": status
    })
```

```python
# รายการสถานะการชำระเงิน
payment_statuses = ["ชำระแล้ว", "ชำระแล้ว", "ชำระแล้ว", "รอชำระเงิน"]
```

```python
records_list = df_medical_records.to_dict('records')
bills_list = []
```

```python
# สุ่ม
for i, rec in enumerate(records_list, 1):
    # รหัสบิล
    bill_id = f"B{i:04d}"

    # ดึงรหัสรักษา ผู้ป่วย
    record_id = rec.get("รหัสรักษา")
    p_id = rec.get("รหัสผู้ป่วย")

    # สุ่มราคาชำระเงิน
    treatment_fee = float(random.choice([200, 300, 400, 500]))

    # กำหนดค่ายา
    meds = rec.get("ยาที่จะสั่ง", "")
    if "Paracetamol" in meds:
        medication_fee = float(random.randint(50, 150))
    elif "Aspirin" in meds or "Ventolin" in meds:
        medication_fee = float(random.randint(250, 600))
    else:
        medication_fee = float(random.randint(300, 1200))

    # คำนวณราคารวม
    total_payment = round(treatment_fee + medication_fee, 2)

    # สถานะการจ่ายเงิน
    status = random.choice(payment_statuses)

    bills_list.append({
        "รหัสใบแจ้งหนี้": bill_id,
        "รหัสการรักษา": record_id,
        "รหัสผู้ป่วย": p_id,
        "ค่าบริการ": treatment_fee,
        "ค่ายา": medication_fee,
        "ยอดชำระรวม": total_payment,
        "สถานะการชำระเงิน": status
    })
```
